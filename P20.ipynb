**Aim:** Create various types of plots using Matplotlib: line plot, bar chart, histogram, pie chart, scatter plot using subplots and figure properties.

### 1. Load Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('in-vehicle-coupon-recommendation.csv')
df.head(2)
```

### 2. Line Plot, Bar Chart, Histogram, Pie Chart, Scatter Plot using Subplots & Figure Properties

```python
# Prepare aggregated/sample data
age_order = ['below21', '21', '26', '31', '36', '41', '46', '50plus']
age_accept = df.groupby('age')['Y'].mean().reindex(age_order) * 100
coupon_counts = df['coupon'].value_counts()
accept_counts = df['Y'].map({1: 'Accepted', 0: 'Rejected'}).value_counts()
sample_df = df.sample(300, random_state=42)

# Create 2x3 Subplots with Figure Properties
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
fig.suptitle('Matplotlib Subplots: Various Plot Types', fontsize=14, fontweight='bold')

# 1. Line Plot
axes[0, 0].plot(age_accept.index, age_accept.values, color='blue', marker='o')
axes[0, 0].set_title('1. Line Plot')
axes[0, 0].set_xlabel('Age Group')
axes[0, 0].set_ylabel('Acceptance Rate (%)')
axes[0, 0].tick_params(axis='x', rotation=30)
axes[0, 0].grid(True)

# 2. Bar Chart
axes[0, 1].bar(coupon_counts.index, coupon_counts.values, color='coral')
axes[0, 1].set_title('2. Bar Chart')
axes[0, 1].tick_params(axis='x', rotation=30)

# 3. Histogram
axes[0, 2].hist(df['temperature'], bins=10, color='lightgreen', edgecolor='black')
axes[0, 2].set_title('3. Histogram')
axes[0, 2].set_xlabel('Temperature (F)')

# 4. Pie Chart
axes[1, 0].pie(accept_counts, labels=accept_counts.index, autopct='%1.1f%%', colors=['skyblue', 'salmon'])
axes[1, 0].set_title('4. Pie Chart')

# 5. Scatter Plot (small jitter added since temperature has only 3 distinct values)
rng = np.random.default_rng(42)
axes[1, 1].scatter(sample_df['temperature'] + rng.normal(0, 1.5, len(sample_df)),
                   sample_df['Y'] + rng.normal(0, 0.03, len(sample_df)),
                   color='purple', alpha=0.5)
axes[1, 1].set_title('5. Scatter Plot')
axes[1, 1].set_xlabel('Temperature (F)')
axes[1, 1].set_ylabel('Accepted (0/1)')

# Remove unused 6th subplot
fig.delaxes(axes[1, 2])

plt.tight_layout()
plt.show()
```
