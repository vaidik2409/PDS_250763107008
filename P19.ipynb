**Aim:** Apply data preprocessing techniques: slicing, filtering, sorting, concatenating, aggregating and normalizing data using Pandas.

### 1. Load Dataset

```python
import pandas as pd

df = pd.read_csv('in-vehicle-coupon-recommendation.csv')
print('Dataset Shape:', df.shape)
df.head(3)
```

### 2. Slicing & Filtering

```python
# Slicing with .iloc and .loc
print('Positional Slicing (.iloc):\n', df.iloc[0:3, 0:4])
print('\nLabel Slicing (.loc):\n', df.loc[0:2, ['destination', 'coupon', 'temperature']])

# Filtering
filtered = df[(df['temperature'] > 70) & (df['Y'] == 1)]
print(f'\nFiltered Count (Temperature > 70 & Y == 1): {len(filtered)}')
```

### 3. Sorting & Concatenating

```python
# Sorting
print('Sorted by Temperature (Descending):\n', df[['destination', 'coupon', 'temperature']].sort_values(by='temperature', ascending=False).head(3))

# Concatenating (Vertical and Horizontal)
v_concat = pd.concat([df.iloc[:2, :3], df.iloc[2:4, :3]], axis=0)
print('\nVertical Concatenation:\n', v_concat)

h_concat = pd.concat([df[['temperature']].head(3), df[['Y']].head(3)], axis=1)
print('\nHorizontal Concatenation:\n', h_con cat)
```
