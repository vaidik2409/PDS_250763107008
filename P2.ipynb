```python
list1 = [1, 2, 3, 4, 5, 3]
list1.append(6)
print(f"list after append 6 = {list1}")
list2 = list1.copy()
print(list2)
list1.pop()
print(list1)
print(list2)
print(list1.count(3))
print(list1.extend([10, 11]))
print(len(list1))
print(list1.index(3))
list1.insert(3, 10)
print(list1)
list1.remove(10)
print(list1)
list1.reverse()
print(list1)
list1.sort()
print(list1)
list1.sort(reverse=True)
print(list1)
```

```python
tuple1 = (1, 2, 3, 4, 3)
print(tuple1.count(3))
print(tuple1.index(2))
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print(set1)
print(set2)

print(set1.union(set2))

print(set1.intersection(set2))

print(set1.difference(set2))
print(set2.difference(set1))

print(set1.symmetric_difference(set2))

print(set1.issubset(set2))
print({1, 2}.issubset(set1))

print(set1.issuperset(set2))
print(set1.issuperset({1, 2}))

set3 = {9, 10}
print(set1.isdisjoint(set2))
print(set1.isdisjoint(set3))
```
