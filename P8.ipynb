```python
# Read data and display top 5 and bottom 5 records

import pandas as pd
import numpy as np
     
df = pd.read_csv("employee.csv")

print("Top 5 Records:")
display(df.head())

print("\nBottom 5 Records:")
display(df.tail())
```

```python
# select column with specific datatype(df.select_dtypes("//datatype"))

print("Numerical Columns:")
display(df.select_dtypes(include="number").head())

print("Categorical Columns:")
display(df.select_dtypes(include="object").head())
```

```python
# findout the maximum an minimum monthly income from the given dataset

print("Maximum Monthly Income:", df["MonthlyIncome"].max())
print("Minimum Monthly Income:", df["MonthlyIncome"].min())
```

```python
# calculate measure of central tendency for appropriate column

income = df["MonthlyIncome"]

print("Mean:", income.mean())
print("Median:", income.median())
print("Mode:", income.mode()[0])
```

```python
# calculate measures of spread

income = df["MonthlyIncome"]

print("Variance:", income.var())
print("Standard Deviation:", income.std())
print("Range:", income.max() - income.min())
print("Q1:", income.quantile(0.25))
print("Q2:", income.quantile(0.50))
print("Q3:", income.quantile(0.75))
print("IQR:", income.quantile(0.75) - income.quantile(0.25))
```

```python
# calculate correlation coefficient

correlation = df.select_dtypes(include="number").corr()

display(correlation)
```
