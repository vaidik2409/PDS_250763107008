**Aim:** Use Pandas to create Series and DataFrames, perform data selection, filtering, grouping and aggregation operations.

### 1. Creating Series and DataFrame

```python
import pandas as pd

# Series
monthly_fees = pd.Series([1500, 2500, 4000], index=["Basic", "Standard", "Premium"])
print("Membership Fees Series:\n", monthly_fees)

# DataFrame
gym_members = {
    "MemberID": [1, 2, 3, 4, 5, 6],
    "Name": ["Tarun", "Kavita", "Sameer", "Aditi", "Nikhil", "Megha"],
    "Plan": ["Standard", "Premium", "Basic", "Premium", "Standard", "Basic"],
    "Age": [24, 29, 22, 31, 27, 21],
    "Months": [6, 12, 3, 12, 6, 3],
    "FeePaid": [15000, 48000, 4500, 48000, 15000, 4500]
}

df = pd.DataFrame(gym_members)
print("\nGym Membership DataFrame:\n", df)
```

### 2. Data Selection

```python
# Column Selection
print("Member Names:\n", df["Name"])
print("\nName & Plan:\n", df[["Name", "Plan"]])

# iloc selection
print("\nFirst 3 rows (iloc):\n", df.iloc[:3])
print("\nIndex 4, Cols 1 & 5 (iloc):\n", df.iloc[4, [1, 5]])

# loc selection
print("\nRows 1 to 4 with Name & Age (loc):\n", df.loc[1:4, ["Name", "Age"]])
```

### 3. Data Filtering

```python
# Members older than 25
print("Members Age > 25:\n", df[df["Age"] > 25])

# Premium members with 12 months duration
print("\nAnnual Premium Members:\n", df[(df["Plan"] == "Premium") & (df["Months"] == 12)])
```

### 4. Grouping and Aggregation

```python
# Average Age per plan
print("Average Age by Plan:\n", df.groupby("Plan")["Age"].mean())

# Plan-wise Aggregation
print("\nPlan Statistics:\n", df.groupby("Plan").agg({
    "FeePaid": ["sum", "mean"],
    "Age": "mean",
    "MemberID": "count"
}))
```
