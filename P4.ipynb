### 1. User-Defined Functions and Scope

```python
# Function with no arguments and no return value
def greet():
    print("Welcome to Python Programming")

greet()

# Function with arguments and return value
def add(a, b):
    return a + b

print("Addition result:", add(15, 25))

# Function returning multiple values
def calc_operations(a, b):
    return a + b, a - b, a * b, a / b

s, d, m, div = calc_operations(20, 5)
print(f"Sum: {s}, Difference: {d}, Product: {m}, Division: {div}")

# Local and Global Scope
count = 10

def update_count():
    global count
    count += 5

update_count()
print("Global count after update:", count)
```

### 2. Different Argument Types

```python
# 1. Positional and Default Arguments
def student_info(name, branch="Computer Engineering", semester=5):
    print(f"Student: {name}, Branch: {branch}, Sem: {semester}")

student_info("Amit")
student_info("Pooja", "Information Technology", 6)

# 2. Keyword Arguments
def employee(name, role, salary):
    print(f"Name: {name}, Role: {role}, Salary: {salary}")

employee(salary=50000, name="Rahul", role="Developer")

# 3. Variable-Length Arguments (*args)
def calculate_total(*numbers):
    return sum(numbers)

print("Total sum with *args:", calculate_total(10, 20, 30, 40, 50))

# 4. Keyword Variable-Length Arguments (**kwargs)
def display_details(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

display_details(name="Karan", roll=101, city="Surat")
```

### 3. Math Module

```python
import math

# Arithmetic, Power and Roots
print("Square root of 25:", math.sqrt(25))
print("Integer square root of 26:", math.isqrt(26))
print("Power (2^4):", math.pow(2, 4))
print("Absolute value of -7.5:", math.fabs(-7.5))

# Number Theory and Combinatorics
print("Factorial of 5:", math.factorial(5))
print("GCD of 24 and 36:", math.gcd(24, 36))
print("LCM of 12 and 15:", math.lcm(12, 15))
print("Combinations (5 choose 2):", math.comb(5, 2))

# Rounding and Floating Operations
print("Ceil of 4.3:", math.ceil(4.3))
print("Floor of 4.7:", math.floor(4.7))
print("Trunc of 4.9:", math.trunc(4.9))
print("fmod(17, 5):", math.fmod(17, 5))

# Logarithms and Constants
print("Log base 10 of 100:", math.log10(100))
print("Natural log of e:", math.log(math.e))
print("Value of Pi:", math.pi)
print("Value of e:", math.e)

# Trigonometry and Geometry
print("sin(90 deg):", math.sin(math.radians(90)))
print("cos(0 deg):", math.cos(0))
print("tan(45 deg):", round(math.tan(math.radians(45)), 2))
print("Hypotenuse of (3, 4):", math.hypot(3, 4))
```

### 4. Random Module

```python
import random

random.seed(42)

# Generating Random Numbers
print("Random float [0, 1):", round(random.random(), 3))
print("Random integer [1, 100]:", random.randint(1, 100))
print("Random uniform [5, 15]:", round(random.uniform(5, 15), 2))
print("Random range [10, 50, step=5]:", random.randrange(10, 50, 5))
print("Gaussian distribution (mu=0, sigma=1):", round(random.gauss(0, 1), 3))

# Sequence Operations
fruits = ["Apple", "Banana", "Cherry", "Mango", "Orange"]
print("Original list:", fruits)
print("Random choice:", random.choice(fruits))
print("Random choices (with replacement, k=3):", random.choices(fruits, k=3))
print("Random sample (without replacement, k=2):", random.sample(fruits, 2))

numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print("Shuffled numbers:", numbers)
```

### 5. Statistics Module

```python
import statistics

data = [12, 15, 18, 20, 20, 22, 25, 28, 30]

print("Mean:", statistics.mean(data))
print("Geometric Mean:", statistics.geometric_mean(data))
print("Harmonic Mean:", statistics.harmonic_mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Multimode:", statistics.multimode([1, 2, 2, 3, 3, 4]))
print("Variance:", statistics.variance(data))
print("Stdev:", statistics.stdev(data))
print("Pop Variance:", statistics.pvariance(data))
print("Pop Stdev:", statistics.pstdev(data))
```
