**Aim:** Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets.

### 1. Statistical Measures using SciPy

```python
from scipy import stats
import numpy as np

# Create Dataset (Order Values in Dollars)
orders = [120, 150, 180, 180, 210, 240, 290, 310]
print("Dataset:", orders)

# Measures of Central Tendency (Mean, Median, Mode)
print("Mean               :", stats.tmean(orders))
print("Median             :", np.median(orders))
result = stats.mode(orders)
print(f"Mode               : {result.mode} (count: {result.count})")

# Measures of Dispersion & Shape (Std Dev, Variance, Skewness, Kurtosis)
print("Standard Deviation :", round(stats.tstd(orders), 4))
print("Variance           :", round(stats.tvar(orders), 4))
print("Skewness           :", round(stats.skew(orders), 4))
print("Kurtosis           :", round(stats.kurtosis(orders), 4))

# Overall SciPy Describe Summary
print("\nSciPy Describe Summary:\n", stats.describe(orders))
```

### 2. Probability Distributions (Normal Distribution)

```python
# Standard Normal Distribution: Probability Density Function (PDF) & Cumulative Distribution Function (CDF)
x = 0.5
print("Standard Normal PDF at x=0.5 :", round(stats.norm.pdf(x, loc=0, scale=1), 4))
print("Standard Normal CDF at Z=2.58:", round(stats.norm.cdf(2.58, loc=0, scale=1), 4))
```

### 3. Hypothesis Testing: One-Sample t-Test

```python
# Testing if mean order value meets sales benchmark (mu=200)
t_stat, p_val = stats.ttest_1samp(orders, popmean=200)

print("Sample:", orders)
print("t-statistic:", round(t_stat, 4))
print("p-value    :", round(p_val, 4))

if p_val < 0.05:
    print("Conclusion : Reject Null Hypothesis (Order value differs significantly from $200).")
else:
    print("Conclusion : Fail to Reject Null Hypothesis (Order value is consistent with $200).")
```

### 4. Correlation Test (Pearson Correlation)

```python
# Pearson Correlation Test (Discount Offered % vs Units Sold)
discount = [5, 10, 15, 20, 25]
units_sold = [12, 18, 25, 34, 45]

result_corr = stats.pearsonr(discount, units_sold)
print("Correlation : ", round(result_corr.statistic, 4))
print("p-value     : ", round(result_corr.pvalue, 6))
```
