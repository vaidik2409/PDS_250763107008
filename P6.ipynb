**Aim:** Demonstrate exception handling using try, except, else, finally blocks and raise statement with custom exceptions.

### 1. Try, Except, Else, and Finally

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    res = a / b
except ZeroDivisionError:
    print("Error: Division by zero")
else:
    print("Result:", res)
finally:
    print("Execution completed")
```

### 2. Multiple Exceptions

```python
try:
    num = int(input("Enter an integer: "))
    res = 100 / num
    print("100 / num =", res)
except ValueError:
    print("Error: Invalid integer input")
except ZeroDivisionError:
    print("Error: Cannot divide by zero")
```

### 3. Raise Statement

```python
try:
    age = int(input("Enter your age: "))
    if age < 18:
        raise ValueError("Age must be 18 or above")
    print("Eligible, age:", age)
except ValueError as e:
    print("Caught Exception:", e)
```

### 4. Custom Exceptions

```python
class NegativeNumberError(Exception):
    pass

try:
    n = int(input("Enter a positive number: "))
    if n < 0:
        raise NegativeNumberError(f"{n} is negative")
    print("Valid number:", n)
except NegativeNumberError as e:
    print("Custom Exception Caught:", e)
```
