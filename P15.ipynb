```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

# Problem:
# A multiple choice quiz has 10 questions with 4 options each (p = 0.25).
# Find the probability of guessing exactly 3 correct answers.
# Then calculate and visualize the Binomial distribution.
n = 10
p = 0.25

probability = binom.pmf(3, n, p)

print("Probability of getting exactly 3 correct answers:", probability)

# Values of X
x = np.arange(0, 11)

# Binomial probabilities
y = binom.pmf(x, n, p)

print("Binomial Distribution")
for i in range(len(x)):
    print("Number of Correct Answers =", x[i], "Probability =", round(y[i], 4))
```

```python
# Problem:
# An emergency room admits an average of 6 patients per hour.
# Find the probability of admitting exactly 5 patients in one hour.
# Then calculate and visualize the Poisson distribution.
lam = 6

probability = poisson.pmf(5, lam)

print("Probability of getting exactly 5 patients:", probability)

# Values of X
x = np.arange(0, 15)

# Poisson probabilities
y = poisson.pmf(x, lam)
print("Poisson Distribution")
for i in range(len(x)):
    print("Number of Patients =", x[i], "Probability =", round(y[i], 4))
```

```python
plt.figure(figsize=(10, 4))

# Binomial plot
plt.subplot(1, 2, 1)
plt.bar(np.arange(0, 11), binom.pmf(np.arange(0, 11), 10, 0.25), color="seagreen")
plt.title("Binomial Distribution")
plt.xlabel("Number of Correct Answers")
plt.ylabel("Probability")

# Poisson plot
plt.subplot(1, 2, 2)
plt.bar(np.arange(0, 15), poisson.pmf(np.arange(0, 15), 6), color="indianred")
plt.title("Poisson Distribution")
plt.xlabel("Number of Patients")
plt.ylabel("Probability")

plt.tight_layout()
plt.show()
```
