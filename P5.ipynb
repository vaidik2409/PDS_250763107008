## Perform read, write and append operations on text files, csv files and binary files using python

```python
with open("temp.txt", "w") as file:
    file.write("Hello world\n")

with open("temp.txt", "r") as file:
    print(file.read())

with open("temp.txt", "a") as file:
    file.write("2nd line.")

with open("temp.txt", "r") as file:
    print(file.read())
```

```python
import csv

data = [
    [1, 2, 3],
    [4, 5, 6]
]

with open("temp.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

with open("temp.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)

with open("temp.csv", "a", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

with open("temp.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
with open("temp.bin", "wb") as file:
    binary_data = b"Hello world"
    file.write(binary_data)

with open("temp.bin", "rb") as file:
    print(file.read())

with open("temp.bin", "wb") as file:
    binary_data = b"2nd line"
    file.write(binary_data)

with open("temp.bin", "rb") as file:
    print(file.read())
```
