```python
!pip install mysql-connector-python
import mysql.connector

connection = mysql.connector.connect(
    host="localhost",
    user="root",
)
cursor = connection.cursor()
```

```python
cursor.execute("CREATE DATABASE IF NOT EXISTS company_db")
cursor.execute("USE company_db")
cursor.execute("""
    CREATE TABLE IF NOT EXISTS employees (
        id INT AUTO_INCREMENT PRIMARY KEY,
        name VARCHAR(100),
        department VARCHAR(50),
        salary INT
    )
""")
```

```python
insert_query = "INSERT INTO employees (name, department, salary) VALUES (%s, %s, %s)"
employee_data = ("John Doe", "Engineering", 75000)
cursor.execute(insert_query, employee_data)
connection.commit()
```

```python
cursor.execute("SELECT * FROM employees")
records = cursor.fetchall()
for row in records:
    print(row)
```

```python
update_query = "UPDATE employees SET salary = %s WHERE name = %s"
update_data = (80000, "John Doe")
cursor.execute(update_query, update_data)
connection.commit()
```

```python
delete_query = "DELETE FROM employees WHERE name = %s"
delete_data = ("John Doe",)
cursor.execute(delete_query, delete_data)
connection.commit()
```

```python
cursor.close()
connection.close()
```
