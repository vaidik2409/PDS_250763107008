# NumPy and Pandas
Working with arrays and dataframes

```python
import numpy as npimport pandas as pd
```

## NumPy Arrays

```python
arr1 = np.array([10, 20, 30, 40, 50])arr2 = np.array([5, 4, 3, 2, 1])print("Array 1 :", arr1)print("Array 2 :", arr2)
```

```python
print("Addition :", arr1 + arr2)print("Subtraction :", arr1 - arr2)print("Multiplication :", arr1 * arr2)print("Division :", arr1 / arr2)
```

```python
print("Dimension :", arr1.ndim)print("Shape :", arr1.shape)print("Size :", arr1.size)print("Data Type :", arr1.dtype)
```

## Matrices

```python
matrix = np.array([    [10,20,30],    [40,50,60],    [70,80,90]])print(matrix)
```

```python
print(matrix[0,0])print(matrix[2,1])
```

```python
print(matrix[0:2,1:3])
```

```python
print("Zeros Matrix")print(np.zeros((3,3)))print("\nOnes Matrix")print(np.ones((2,4)))print("\nIdentity Matrix")print(np.eye(4))
```

## Random Numbers and Statistics

```python
random_array = np.random.randint(1,100,10)print(random_array)
```

```python
print("Maximum :", np.max(random_array))print("Minimum :", np.min(random_array))print("Sum :", np.sum(random_array))print("Mean :", np.mean(random_array))print("Median :", np.median(random_array))print("Standard Deviation :", np.std(random_array))
```

```python
numbers = np.arange(1,13)print("Original")print(numbers)print("\nReshape")print(numbers.reshape(3,4))
```

## Pandas Series

```python
series = pd.Series([100,200,300,400],                   index=["A","B","C","D"])print(series)
```

## DataFrame

```python
student = {    "Name":["Vaidik", "Dhruv", "Kunal", "Rohan", "Sahil"],    "Age":[21, 20, 22, 19, 21],    "Marks":[90, 74, 82, 68, 95],    "City":["Indore", "Bhopal", "Gwalior", "Ujjain", "Jabalpur"]}df = pd.DataFrame(student)print(df)
```

```python
print("Shape :", df.shape)print("Columns :", df.columns.tolist())print("\nData Types")print(df.dtypes)
```

```python
print("First Three Rows")print(df.head(3))print("\nLast Two Rows")print(df.tail(2))
```

```python
print(df["Name"])print("\nMarks")print(df["Marks"])
```

```python
print("Students having Marks > 80")print(df[df["Marks"]>80])
```

```python
print("Sort by Marks")print(df.sort_values("Marks"))
```

```python
df["Result"] = ["Pass","Pass","Pass","Pass","Pass"]print(df)
```

```python
print("Average Marks")print(df["Marks"].mean())print("\nHighest Marks")print(df["Marks"].max())print("\nLowest Marks")print(df["Marks"].min())
```

```python
print("Average Marks City Wise")print(df.groupby("City")["Marks"].mean())
```

## Handling Missing Values

```python
df.loc[2,"Marks"] = np.nanprint(df)print("\nMissing Values")print(df.isnull().sum())
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())print(df)
```

```python
df.to_csv("student_data.csv", index=False)print("CSV File Created Successfully")
```
