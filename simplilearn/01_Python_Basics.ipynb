## Data Structures

```python
marks = [78, 85, 90, 88, 95]print("Original List")print(marks)marks.append(80)marks.insert(2, 100)print("Updated List")print(marks)print("Maximum =", max(marks))print("Minimum =", min(marks))print("Sum =", sum(marks))print("Average =", sum(marks)/len(marks))
```

```python
colors = ("Red", "Blue", "Green", "Black")print(colors)print("First =", colors[0])
```

```python
numbers = {10,20,30,20,10,40}print(numbers)numbers.add(50)print(numbers)
```

## Control Flow

```python
marks = 82if marks >= 90:    print("Grade A")elif marks >= 75:    print("Grade B")elif marks >= 50:    print("Grade C")else:    print("Fail")
```

```python
for i in range(1,6):    print(i)
```

```python
count = 1while count <= 5:    print(count)    count += 1
```
