# Data Visualization
Using Matplotlib and Seaborn

```python
import pandas as pdimport matplotlib.pyplot as pltimport seaborn as sns
```

## Create Sample Dataset

```python
data = {    "Student": ["Vaidik", "Dhruv", "Kunal", "Rohan", "Sahil", "Nikhil", "Parth"],    "Maths": [90, 74, 82, 68, 95, 77, 86],    "Science": [88, 78, 85, 72, 93, 80, 89],    "English": [85, 70, 80, 65, 92, 76, 84]}df = pd.DataFrame(data)print(df)
```

```python
print("Shape :", df.shape)print("\nColumns")print(df.columns)print("\nSummary Statistics")print(df.describe())
```

## Matplotlib Charts

```python
plt.figure(figsize=(8,5))plt.plot(df["Student"], df["Maths"], marker='o')plt.title("Maths Marks")plt.xlabel("Students")plt.ylabel("Marks")plt.grid(True)plt.show()
```

```python
plt.figure(figsize=(8,5))plt.bar(df["Student"], df["Science"])plt.title("Science Marks")plt.xlabel("Students")plt.ylabel("Marks")plt.show()
```

```python
plt.figure(figsize=(8,5))plt.scatter(df["Maths"], df["Science"])plt.title("Maths vs Science")plt.xlabel("Maths")plt.ylabel("Science")plt.grid(True)plt.show()
```

```python
plt.figure(figsize=(7,5))plt.hist(df["English"], bins=5)plt.title("English Marks Distribution")plt.xlabel("Marks")plt.ylabel("Frequency")plt.show()
```

```python
plt.figure(figsize=(6,6))plt.pie(    df["Maths"],    labels=df["Student"],    autopct="%1.1f%%",    startangle=90)plt.title("Maths Marks Percentage")plt.show()
```

```python
plt.figure(figsize=(6,5))plt.boxplot(df["Science"])plt.title("Science Marks Box Plot")plt.ylabel("Marks")plt.show()
```

## Seaborn Charts

```python
sns.lineplot(x="Student", y="English", data=df)plt.title("English Marks")plt.show()
```

```python
sns.barplot(x="Student", y="Maths", data=df)plt.title("Maths Bar Plot")plt.show()
```

```python
sns.scatterplot(    x="Maths",    y="Science",    data=df)plt.title("Maths vs Science")plt.show()
```

```python
sns.histplot(df["Maths"], bins=5)plt.title("Maths Histogram")plt.show()
```

## Correlation and Heatmap

```python
print("Correlation Matrix")print(df[["Maths", "Science", "English"]].corr())
```

```python
sns.heatmap(    df[["Maths","Science","English"]].corr(),    annot=True,    cmap="coolwarm")plt.title("Correlation Heatmap")plt.show()
```

```python
sns.pairplot(    df[["Maths","Science","English"]])plt.show()
```

## Analysis

```python
df["Average"] = (    df["Maths"] +    df["Science"] +    df["English"]) / 3print("Average Marks")print(df[["Student","Average"]])
```

```python
highest = df.loc[df["Average"].idxmax()]print("Top Student")print(highest)
```
