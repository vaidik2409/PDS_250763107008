**Aim:** Perform data cleaning operations: detect and handle missing values, remove duplicate records, and treat outliers in a dataset.

### 1. Load Dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv('in-vehicle-coupon-recommendation.csv')
print('Initial Shape:', df.shape)
df.head(5)
```

### 2. Detect & Handle Missing Values

```python
# Detect missing values
print('Missing values per column:')
print(df.isnull().sum()[df.isnull().sum() > 0])

# 'car' is missing in ~99% of rows -> drop the column instead of imputing it
print("\n'car' is missing in 99.1% of rows, so it is dropped instead of imputed.")
df.drop(columns=['car'], inplace=True)

# Handle remaining missing values by imputing Mode
for col in ['Bar', 'CoffeeHouse', 'CarryAway', 'RestaurantLessThan20', 'Restaurant20To50']:
    df[col] = df[col].fillna(df[col].mode()[0])

print('\nRemaining missing values:', df.isnull().sum().sum())
```

### 3. Detect & Remove Duplicate Records

```python
# Detect duplicates
print('Duplicate records count:', df.duplicated().sum())

# Remove duplicates
df.drop_duplicates(inplace=True)
print('Shape after removing duplicates:', df.shape)
```

### 4. Detect & Treat Outliers

```python
# Detect outliers using IQR Method on 'temperature'
Q1 = df['temperature'].quantile(0.25)
Q3 = df['temperature'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['temperature'] < lower_bound) | (df['temperature'] > upper_bound)]
print(f'IQR: {IQR}, Lower Bound: {lower_bound}, Upper Bound: {upper_bound}')
print('Outliers count in temperature:', len(outliers))

# Treat outliers using Capping (Winsorization)
df['temperature'] = df['temperature'].clip(lower=lower_bound, upper=upper_bound)
print('Outliers treated! Capped Min:', df['temperature'].min(), 'Capped Max:', df['temperature'].max())
```

### 5. Cleaning Summary

```python
print('Final dataset shape:', df.shape)
print('Missing values remaining:', df.isnull().sum().sum())
print('Duplicate records remaining:', df.duplicated().sum())
```
